```python
# Create a dictionary, tuple, and list of students

students_dict = {
    101: "Rahul",
    102: "Priya",
    103: "Amit"
}

students_tuple = ("Rahul", "Priya", "Amit")
students_list = ["Rahul", "Priya", "Amit"]

print("Dictionary:", students_dict)
print("Tuple:", students_tuple)
print("List:", students_list)

# Add a student
students_dict[104] = "Sneha"
print("After adding:", students_dict)

# Delete a student
del students_dict[102]
print("After deleting:", students_dict)

# Update a student's name
students_dict[103] = "Ankit"
print("After updating:", students_dict)
```
